106. To evaluate the following sum:
sum = Σ (-1)^n ((x^(n/2))/n)(n+1) for n = 1 to 10

```python
x = float(input("Enter x: "))
print("You entered:", x)
result = 0.0
for n in range(1, 11):
    term = ((-1) ** n) * ((x ** (n / 2)) / n) * (n + 1)
    result += term
print("Result:", result)
```

107. To find the value of the following formula:
nCr = n!/(r! * (n-r)!)

```python
import math
n = int(input("Enter n: "))
r = int(input("Enter r: "))
if 0 <= r <= n:
    nCr = math.comb(n, r)
    print(f"{n}C{r} =", nCr)
else:
    print("Please enter values with 0 <= r <= n.")
```

109. Read any two positive numbers say n1 & n2. Assume n1>n2.
Print all even numbers that lie between n2 and n1.
Print the total number of even numbers between n1 and n2.

```python
n1 = int(input("Enter n1 (larger number): "))
n2 = int(input("Enter n2 (smaller number): "))
if n1 > n2:
    evens = [str(num) for num in range(n2 + 1, n1) if num % 2 == 0]
    print("Even numbers between", n2, "and", n1, ":", " ".join(evens))
    print("Total even numbers:", len(evens))
else:
    print("Please ensure n1 > n2.")
```

111. To find the total number of odd digits within the given number and print the sum of all odd digits.

```python
number = input("Enter a number: ")
odd_count = 0
odd_sum = 0
for ch in number:
    if ch.isdigit():
        value = int(ch)
        if value % 2 != 0:
            odd_count += 1
            odd_sum += value
print("Total odd digits:", odd_count)
print("Sum of odd digits:", odd_sum)
```

---

112. To find the total number of even digits within the given number and print the sum of all even digits.

```python
number = input("Enter a number: ")
even_count = 0
even_sum = 0
for ch in number:
    if ch.isdigit():
        value = int(ch)
        if value % 2 == 0:
            even_count += 1
            even_sum += value
print("Total even digits:", even_count)
print("Sum of even digits:", even_sum)
```

114. To convert the given octal number into its equivalent decimal number.

119. Read a keystroke line and print the total number of vowels, blank spaces, and digits entered.

```python
line = input("Enter text and press Enter: ")
vowels = sum(1 for ch in line.lower() if ch in 'aeiou')
spaces = line.count(' ')
digits = sum(1 for ch in line if ch.isdigit())
print("Total vowels:", vowels)
print("Total blank spaces:", spaces)
print("Total digits:", digits)
```

120. Two 4-digit positive integers: calculate the sum of the products of digits in the same positions.

```python
first_number = input("Enter the first 4-digit number: ")
second_number = input("Enter the second 4-digit number: ")
if len(first_number) == 4 and len(second_number) == 4 and first_number.isdigit() and second_number.isdigit():
    total = sum(int(first_number[i]) * int(second_number[i]) for i in range(4))
    print("Sum of products of corresponding digits:", total)
else:
    print("Please enter two valid 4-digit positive integers.")
```

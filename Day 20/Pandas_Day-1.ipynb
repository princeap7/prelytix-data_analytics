## 1. Introduction to Pandas 

Pandas is the primary Python library for structured/tabular data. Built on NumPy, it adds labelled axes, 
mixed data types, and a huge toolset for data cleaning, reshaping, and analysis. 

## 1.1 Why Pandas? 
- Python lists and NumPy lack column labels and mixed-type support 
- Pandas Series and DataFrame mirror spreadsheets but with code power 
- Reads CSV, Excel, JSON, SQL, Parquet — all with a single function call 
- Integrates directly with Matplotlib, Seaborn, and Scikit-learn 
- Handles real-world messy data: nulls, mixed types, duplicates, dates

## 2. Pandas Series 
A Series is a 1-dimensional labelled array. Think of it as a single column with an index. 

## 2.1 Creating Series

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40])

print("Series created from list:")
print(s)

marks = pd.Series([88, 75, 92, 60], index=['Math', 'Sci', 'Eng', 'Hist'])

print("Marks Series:")
print(marks)

print("Marks in Math:")
print(marks['Math'])

print("Subjects with marks greater than 80:")
print(marks[marks > 80])

d = pd.Series({'a': 1, 'b': 2, 'c': 3})

print("Series created from dictionary:")
print(d)

s1 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
s2 = pd.Series([10, 20, 30], index=['b', 'c', 'd'])

print("Addition of two Series with aligned indices:")
print(s1 + s2)

print("Values of marks Series:")
print(marks.values)

print("Index of marks Series:")
print(marks.index)

print("Data type of marks Series:")
print(marks.dtype)
```

```python
df = pd.DataFrame(
    np.arange(12).reshape(4, 3),
    index=['a', 'b', 'c', 'd'],
    columns=['X', 'Y', 'Z']
)

print("Original DataFrame:")
print(df)

print("Value at row 'b' and column 'Y':")
print(df.loc['b', 'Y'])

print("Rows a to c and columns X to Y:")
print(df.loc['a':'c', 'X':'Y'])

print("Value at row index 1 and column index 1:")
print(df.iloc[1, 1])

print("Rows 0 to 2 and columns 0 to 1:")
print(df.iloc[0:3, 0:2])

print("Value using .at:")
print(df.at['b', 'Y'])

print("Value using .iat:")
print(df.iat[1, 1])
```

## 5. Missing Values 
## 5.1 Detecting & Handling

```python
df = sns.load_dataset('titanic')

print("Missing values per column:")
print(df.isnull().sum())

print("Percentage of missing values:")
print(df.isnull().sum() / len(df) * 100)

clean = df.dropna()

print("Shape after dropping rows with any null values:")
print(clean.shape)

df2 = df.dropna(subset=['age', 'embarked'])

print("Shape after dropping rows with null age or embarked:")
print(df2.shape)

df['age'] = df['age'].fillna(df['age'].median())

df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

df['deck'] = df['deck'].cat.add_categories('Unknown')
df['deck'] = df['deck'].fillna('Unknown')

print("Missing values after filling:")
print(df[['age', 'embarked', 'deck']].isnull().sum())

ts = pd.DataFrame({
    'val': [1, np.nan, np.nan, 4]
})

print("Original time series:")
print(ts)

print("Forward fill:")
print(ts.ffill())

print("Backward fill:")
print(ts.bfill())

print("Linear interpolation:")
print(ts.interpolate())
```

## 5.2 Tasks — Missing Values 
Task 1: On the Titanic dataset: (a) find % missing per column, (b) drop columns with >50% missing, (c) fill Age with median grouped by Pclass. 

Task 2: Create a DataFrame with intentional NaN values. Compare dropna(), fillna(0), fillna(mean), and interpolate() — show the result of each. 

Task 3: Use df.fillna(df.mean()) to fill all numeric nulls at once. Then use df.fillna(df.mode().iloc[0]) for non-numeric nulls. 

```python
print("Task 1")

df = sns.load_dataset('titanic')

print("Percentage of missing values per column:")
print(df.isnull().sum() / len(df) * 100)

missing_percent = df.isnull().mean() * 100

df = df.loc[:, missing_percent <= 50]

print("Columns remaining after removing columns with >50% missing:")
print(df.columns)

df['age'] = df.groupby('pclass')['age'].transform(
    lambda x: x.fillna(x.median())
)

print("Age column filled using median grouped by pclass.")
```

```python
print("Task 2")

data = pd.DataFrame({
    'A': [10, np.nan, 30, np.nan, 50],
    'B': [5, 15, np.nan, 35, 45]
})

print("Original DataFrame:")
print(data)

print("After dropna():")
print(data.dropna())

print("After fillna(0):")
print(data.fillna(0))

print("After fillna(mean):")
print(data.fillna(data.mean()))

print("After interpolate():")
print(data.interpolate())
```

```python
print("Task 3")

data = pd.DataFrame({
    'A': [10, np.nan, 30],
    'B': [5, 15, np.nan],
    'C': ['X', np.nan, 'Y']
})

print("Original DataFrame:")
print(data)

data = data.fillna(data.mean(numeric_only=True))
data = data.fillna(data.mode().iloc[0])

print("DataFrame after filling numeric and non-numeric nulls:")
print(data)
```

## 6. Column Operations

```python
df = sns.load_dataset('titanic').copy()

df['family_size'] = df['sibsp'] + df['parch'] + 1

df['alone'] = (df['family_size'] == 1).astype(int)

df['age_group'] = df['age'].apply(
    lambda x: 'Child' if x < 18 else 'Senior' if x >= 60 else 'Adult'
)

df.drop(columns=['deck', 'embark_town'], inplace=True)

df.rename(
    columns={
        'pclass': 'Pclass',
        'survived': 'Survived'
    },
    inplace=True
)

df['age'] = df['age'].fillna(df['age'].median()).astype(int)

print("New columns:")
print(df[['family_size', 'alone', 'age_group']].head())
```

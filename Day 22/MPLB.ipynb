## 1. Introduction to Matplotlib 
Matplotlib is the fundamental data visualization library in Python. It provides both a quick Pyplot 
interface and a robust Object-Oriented interface. 

## 1.1 Line and Scatter Plots 

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x,y1,color='blue',linestyle='-',linewidth=2,label='Sine')
x_scatter = np.random.rand(50) * 10
y_scatter = np.random.rand(50) * 10
plt.scatter(x_scatter,y_scatter,color='red',marker='x',s=50,label='Random Points')
plt.legend()
plt.show()
```

## 2. Categorical & Distribution Plots 
Bar charts compare categorical data. Histograms show the distribution of a single numerical variable. Pie charts show proportions. 

## 2.1 Bar, Histogram, and Pie 

```python
categories = ['Apples', 'Bananas', 'Cherries']

counts = [40, 25, 15]

plt.bar(categories,counts,color=['red', 'yellow', 'darkred'])

plt.title("Fruit Inventory")

plt.show()

data = np.random.randn(1000)

plt.hist(data,bins=30,edgecolor='black',color='skyblue')

plt.title("Data Distribution")

plt.show()

plt.pie(counts,labels=categories,autopct='%1.1f%%',explode=[0.1, 0, 0])

plt.title("Fruit Share")

plt.show()
```

## 3. Customizing & Subplots 
A good plot requires clear labels, titles, grids, and proper axis limits. Subplots allow multiple charts in one 
figure.

## 3.1 Titles, Labels, Limits, Grid 

```python
plt.plot([1, 2, 3],[10, 20, 15],label='Sales')
plt.title("Q1 Sales Report",fontsize=16,fontweight='bold')
plt.xlabel("Month",fontsize=12)
plt.ylabel("Revenue ($)",fontsize=12)
plt.xlim(0, 4)
plt.ylim(0, 25)
plt.xticks([1, 2, 3],['Jan', 'Feb', 'Mar'])
plt.grid(True,linestyle='--',alpha=0.5)
plt.legend()
plt.show()
```

## 3.2 Subplots 

```python
fig, axes = plt.subplots(nrows=1,ncols=2,figsize=(10, 4))
axes[0].plot([1, 2, 3],[3, 2, 1])
axes[0].set_title('Descending')
axes[1].plot([1, 2, 3],[1, 2, 3])
axes[1].set_title('Ascending')
plt.tight_layout()
plt.savefig('my_subplots.png',dpi=300)
plt.show()
```

## 3.3 Tasks — Customization & Subplots

Task 1: Take your line plot from Task 1.1 and add a title, x/y labels, a grid, and set x-axis limits from 0 to 12.

Task 2: Create a 2x2 subplot grid. Plot a line, scatter, bar, and histogram in the 4 axes respectively. Add titles to each.

```python
print("Task 1")

x = np.arange(1, 11)

y = x ** 2

plt.plot(x,y,linestyle='--',color='green')
plt.title("Squares of Numbers")
plt.xlabel("Number")
plt.ylabel("Square")
plt.xlim(0, 12)
plt.grid()
plt.show()
```

```python
print("Task 2")
fig, axes = plt.subplots(2,2,figsize=(8, 6))
axes[0, 0].plot([1, 2, 3], [1, 4, 9])
axes[0, 0].set_title("Line Plot")
axes[0, 1].scatter([1, 2, 3, 4],[5, 7, 6, 8])
axes[0, 1].set_title("Scatter Plot")
axes[1, 0].bar(['A', 'B', 'C'],[10, 15, 8])
axes[1, 0].set_title("Bar Chart")
axes[1, 1].hist(np.random.randn(100))
axes[1, 1].set_title("Histogram")
plt.tight_layout()
plt.show()
```

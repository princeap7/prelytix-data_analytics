## 1. Introduction to Seaborn & Relational Plots 
Seaborn is a statistical plotting library built on Matplotlib. It works perfectly with Pandas DataFrames and produces beautiful charts with minimal code. 

## 1.1 Scatter and Line Plots

```python
import seaborn as sns 
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')

sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', style='smoker', size='size')

plt.title("Tips vs Total Bill")

plt.show()

fmri = sns.load_dataset('fmri')

sns.lineplot(data=fmri, x='timepoint', y='signal', hue='event')

plt.title("FMRI Signal over Time")

plt.show()
```

## 2. Distribution and Categorical Plots 
These plots are critical for understanding how data is spread and comparing different categories. 
## 2.1 Histograms, KDEs, and Categorical Plots 

```python
penguins = sns.load_dataset('penguins')

tips = sns.load_dataset('tips')

sns.histplot(data=penguins, x='body_mass_g', kde=True, hue='species')

plt.title("Body Mass Distribution")

plt.show()

sns.countplot(data=tips, x='day', hue='sex')

plt.title("Count of Customers by Day and Sex")

plt.show()

sns.boxplot(data=tips, x='day', y='total_bill', hue='smoker')

plt.title("Total Bill Distribution by Day")

plt.show()
```

## 3. Matrix Plots & Pairplots 
Heatmaps are perfect for viewing correlation matrices. Pairplots show relationships across all numeric 
variables in a dataset at once. 
## 3.1 Heatmaps and Pairplots

```python
titanic = sns.load_dataset('titanic')

numeric_df = titanic.select_dtypes(include='number')

corr = numeric_df.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title("Titanic Correlation Heatmap")

plt.show()

penguins = sns.load_dataset('penguins')

sns.pairplot(penguins, hue='species')

plt.show()
```

## 3.2 Tasks — Matrix & Pairplot
Task 1: Calculate the correlation matrix for the 'tips' dataset.Plot it as a heatmap with annotations (annot=True) and a 'viridis' colormap. 

Task 2: Create a pairplot for the 'iris' dataset, coloring the points by 'species'. Which species is most easily separable?

```python
print("Task 1")

tips = sns.load_dataset('tips')

numeric_df = tips.select_dtypes(include='number')

corr = numeric_df.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(corr, annot=True, cmap='viridis')

plt.title("Tips Correlation Heatmap")

plt.show()
```

```python
print("Task 2")

iris = sns.load_dataset('iris')

sns.pairplot(iris, hue='species')

plt.show()

print("Most easily separable species: Setosa")
```

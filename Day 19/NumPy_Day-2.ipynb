## 1. Advanced Boolean & Fancy Indexing 
## 1.1 Compound Boolean Conditions 

```python
import numpy as np

data = np.array([15, 42, -3, 87, -12, 55, 0, 33])

pos_small = data[(data > 0) & (data < 50)]
print("Positive values less than 50:", pos_small)

neg_or_big = data[(data < 0) | (data > 80)]
print("Negative values or values greater than 80:", neg_or_big)

not_zero = data[~(data == 0)]
print("Values not equal to zero:", not_zero)

print("Any values less than 0:", np.any(data < 0))
print("All values greater than 0:", np.all(data > 0))
print("All values greater than 0 (axis=0):", np.all(data > 0, axis=0))

idx = np.nonzero(data > 40)
print("Indices of values greater than 40:", idx)
print("Values greater than 40:", data[idx])
```

## 1.2 np.where — Vectorised if-else

```python
scores = np.array([45, 82, 67, 55, 91, 38, 74])

clipped = np.where(scores >= 60, scores, 60)
print("Scores after replacing values below 60:", clipped)

labels = np.where(scores >= 80, 'Dist',
         np.where(scores >= 60, 'Merit', 'Pass'))
print("Performance labels:", labels)

bonus = np.where(scores >= 80, scores * 1.1, scores)
print("Scores after applying bonus:", bonus.astype(int))

conds = [scores >= 80, scores >= 60, scores >= 40]
vals = ['A', 'B', 'C']
grades = np.select(conds, vals, default='D')

print("Grades assigned using np.select():", grades)
```

## 1.3 Fancy Indexing & np.ix_ 

```python
m = np.arange(25).reshape(5, 5)

print("Selected rows (0, 2, 4):")
print(m[[0, 2, 4], :])

r = [0, 1, 2]
c = [4, 3, 2]
print("Elements at specified (row, column) pairs:")
print(m[r, c])

rows = [0, 2]
cols = [1, 3]
sub = m[np.ix_(rows, cols)]

print("Sub-matrix using np.ix_():")
print(sub)

print("Matrix after reversing rows:")
print(m[::-1, :])

print("Matrix after reversing columns:")
print(m[:, ::-1])
```

## 2. Broadcasting — Deep Dive 

Broadcasting allows NumPy to perform operations on arrays of different shapes without copying data, 
following three rules. 

## 2.1 Broadcasting Rules 

- Rule 1: If arrays have different ndim, the shape of the smaller is padded with 1s on the LEFT. 
- Rule 2: Arrays with size 1 along any dimension are stretched to match the other. 
- Rule 3: If sizes differ and neither is 1 → ValueError.

```python
col = np.array([[0], [10], [20]])
row = np.array([1, 2, 3, 4])

print("Result of broadcasting (column + row):")
print(col + row)

data = np.random.randint(1, 100, (4, 5)).astype(float)

print("Original data:")
print(data)

row_means = data.mean(axis=1, keepdims=True)
centred = data - row_means

print("Row means:")
print(row_means)

print("Data after centering:")
print(centred)

print("Mean of each row after centering:")
print(centred.mean(axis=1))

a = np.array([1, 2, 3])[:, np.newaxis]
b = np.array([10, 20, 30, 40])

print("Outer product using broadcasting:")
print(a * b)
```

## 3. Linear Algebra with np.linalg 

## 3.1 Matrix Multiplication 

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix multiplication using np.dot():")
print(np.dot(A, B))

print("Matrix multiplication using @ operator:")
print(A @ B)

print("Matrix multiplication using np.matmul():")
print(np.matmul(A, B))

print("Element-wise multiplication:")
print(A * B)

print("Matrix multiplication:")
print(A @ B)
```

## 3.2 linalg Functions 

```python
A = np.array([[2., 1.], [1., 3.]])

Ai = np.linalg.inv(A)

print("Inverse of matrix A:")
print(Ai)

print("A multiplied by its inverse (Identity matrix):")
print(A @ Ai)

print("Determinant of A:")
print(np.linalg.det(A))

vals, vecs = np.linalg.eig(A)

print("Eigenvalues of A:")
print(vals)

print("Eigenvectors of A:")
print(vecs)

b = np.array([8., 13.])

x = np.linalg.solve(A, b)

print("Solution vector x:")
print(x)

print("Verification (A @ x):")
print(A @ x)

v = np.array([3., 4.])

print("L2 norm of vector v:")
print(np.linalg.norm(v))

print("Frobenius norm of matrix A:")
print(np.linalg.norm(A, 'fro'))
```

```python
print("Task 3.3.1")

A = np.random.randint(1, 10, size=(3, 3)).astype(float)

print("Matrix A:")
print(A)

A_inv = np.linalg.inv(A)

identity_check = A @ A_inv

print("A @ inv(A):")
print(identity_check)

print("Is A @ inv(A) approximately equal to Identity matrix?")
print(np.allclose(identity_check, np.eye(3)))
```

```python
print("Task 3.3.2")

A = np.array([[2, 1],
              [1, 3]])

b = np.array([5, 10])

solution = np.linalg.solve(A, b)

print("Solution [x, y]:")
print(solution)

print("Verification (A @ solution):")
print(A @ solution)
```

```python
print("Task 3.3.3")

A = np.random.randint(1, 10, size=(3, 3))

S = A.T @ A

print("Symmetric matrix S = A.T @ A:")
print(S)

eigenvalues, eigenvectors = np.linalg.eig(S)

print("Eigenvalues of S:")
print(eigenvalues)

print("Are all eigenvalues non-negative?")
print(np.all(eigenvalues >= 0))
```

```python
print("Task 3.3.4")

v1 = np.random.rand(3)
v2 = np.random.rand(3)

v1 = v1 / np.linalg.norm(v1)
v2 = v2 / np.linalg.norm(v2)

dot_product = np.dot(v1, v2)

print("First unit vector:")
print(v1)

print("Second unit vector:")
print(v2)

print("Dot product of the two unit vectors:")
print(dot_product)

print("Interpretation:")
print("A value close to 1 means vectors are similar, close to 0 means nearly perpendicular, and close to -1 means opposite directions.")
```

## 4. Sorting, Searching & Set Operations 
## 4.1 Sorting

```python
arr = np.array([3, 1, 4, 1, 5, 9, 2, 6, 5])

print("Sorted copy of array:")
print(np.sort(arr))

arr.sort()

print("Array after in-place sorting:")
print(arr)

print("Indices that would sort the array:")
print(np.argsort(arr))

m = np.random.randint(1, 20, (3, 4))

print("Original matrix:")
print(m)

print("Row-wise sorted matrix:")
print(np.sort(m, axis=1))

print("Column-wise sorted matrix:")
print(np.sort(m, axis=0))

k = 3
top_idx = np.argsort(arr)[-k:][::-1]

print("Largest 3 elements:")
print(arr[top_idx])
```

## 4.2 Searching & Set Ops

```python
arr = np.array([1, 1, 2, 3, 4, 4, 5])

vals, counts = np.unique(arr, return_counts=True)

print("Unique values:")
print(vals)

print("Count of each unique value:")
print(counts)

print("Index where 3 should be inserted:")
print(np.searchsorted(arr, 3))

a = np.array([1, 2, 3, 4, 5])
b = np.array([3, 4, 5, 6, 7])

print("Common elements:")
print(np.intersect1d(a, b))

print("Union of arrays:")
print(np.union1d(a, b))

print("Elements present only in A:")
print(np.setdiff1d(a, b))

print("Membership of elements of A in B:")
print(np.in1d(a, b))
```

## 6. Performance & Vectorisation 
## 6.1 Vectorisation Patterns

```python
import time

n = 3_000_000

a = np.random.rand(n)
b = np.random.rand(n)

start = time.perf_counter()
c = [a[i] + b[i] for i in range(n)]
print(f"Loop execution time: {time.perf_counter() - start:.3f} seconds")

start = time.perf_counter()
c = a + b
print(f"NumPy execution time: {time.perf_counter() - start:.4f} seconds")

a += b

np.add(a, b, out=a)

result = np.empty(n)

np.multiply(a, b, out=result)

print("Multiplication result stored using pre-allocation.")
```

```python
print("Task 6.3.2")

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

outer_einsum = np.einsum('i,j->ij', a, b)
outer_np = np.outer(a, b)

print("Outer product using einsum:")
print(outer_einsum)

print("Outer product using np.outer:")
print(outer_np)

print("Are both results equal?")
print(np.array_equal(outer_einsum, outer_np))
```

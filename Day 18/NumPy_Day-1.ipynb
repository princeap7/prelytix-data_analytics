## 1.1 NumPy

NumPy (Numerical Python) is the foundational library for scientific computing in Python. It provides a high-performance multidimensional array object called ndarray and a large collection of mathematical functions to operate on these arrays efficiently.

- Python lists are slow for numerical work — they store pointers, not raw values 
- NumPy arrays store homogeneous, typed data in contiguous memory → much faster 
- Vectorised operations eliminate explicit Python loops
- Forms the backbone of Pandas, Matplotlib, Scikit-learn, TensorFlow, SciPy 
- Native support for linear algebra, Fourier transforms, and random sampling

## 4.2 2D Indexing

```python
m = np.arange(1, 13).reshape(3, 4)
print(m)
print(m[0, 2]) # 3 (row 0, col 2)
print(m[1, :]) # [5 6 7 8] row 1
print(m[:, 2]) # [3 7 11] col 2
print(m[0:2, 1:3]) # [[2 3],[6 7]] sub-matrix
print(m[-1, -1]) # 12 last element
print(m[0:2, 1:3]) # [[2 3],[6 7]] sub-matrix
```

## 5 Array Operations & Universal Functions

## 5.1 Element-wise Arithmetic

```python
a = np.array([1, 2, 3, 4]) 
b = np.array([10, 20, 30, 40])
print("Addition:", a + b)
print("Subtraction:", b - a)
print("Multiplication:", a * b)
print("Division:", b / a)
print("Floor Division:", b // a)
print("Modulus:", b % 3)
print("Square:", a ** 2)
```

## 5.2 Universal Functions (ufuncs)

```python
a = np.array([1, 4, 9, 16])

print("Square Root:", np.sqrt(a))
print("\nSquare:", np.square(a))
print("\nAbsolute Values:", np.abs(np.array([-1, -2, 3])))
print("\nExponential:", np.exp(np.array([0, 1, 2])))

print("\nNatural Logarithm:", np.log(a))
print("\nBase-2 Logarithm:", np.log2(a))
print("\nBase-10 Logarithm:", np.log10(a))

print("\nSine Values:", np.sin(np.linspace(0, np.pi, 5)))

print("\nCeiling Values:", np.ceil(np.array([1.2, 3.7])))
print("\nFloor Values:", np.floor(np.array([1.9, 3.1])))

print("\nRounded Values:", np.round(np.array([1.56, 3.44]), 1))
```

## 5.3 Aggregate Functions

```python
m = np.array([[1, 2, 3],[4, 5, 6]])
print("Original Matrix:\n", m)
print("\nSum:", np.sum(m))
print("Sum (axis=0):", np.sum(m, axis=0))
print("Sum (axis=1):", np.sum(m, axis=1))

print("\nMean:", np.mean(m))
print("Standard Deviation:",np.std(m))
print("Variance:", np.var(m))

print("\nMinimum Value:", np.min(m))
print("Maximum Value:", np.max(m))

print("Index of Minimum Value:", np.argmin(m))
print("Index of Maximum Value:", np.argmax(m))

print("\nCumulative Sum (axis=1):", np.cumsum(m, axis=1))

print("Product (axis=0):", np.prod(m, axis=0))
```

### 6.2 Stacking & Splitting
concat/vstack/hstack/stack and splitting with `vsplit`/`hsplit`.

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print('vstack:')
print(np.vstack([a, b]))
print('\nhstack:')
print(np.hstack([a, b]))
print('\nconcatenate axis=0:')
print(np.concatenate([a, b], axis=0))
print('\nstack axis=0 shape:')
print(np.stack([a, b], axis=0).shape)
x = np.arange(12).reshape(4, 3)
print('\nx:')
print(x)
print('\nvsplit into 2:')
print(np.vsplit(x, 2))
print('\nhsplit into 3:')
print(np.hsplit(x, 3))
```
